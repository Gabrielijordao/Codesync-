```
dados = [5, 3, 8, 1, 2]


def bubble_sort(vetor):
    n = len(vetor)

    comparacoes = 0
    trocas = 0

    for i in range(n):
        for j in range(n - 1 - i):

            comparacoes += 1

            if vetor[j] > vetor[j + 1]:
                vetor[j], vetor[j + 1] = vetor[j + 1], vetor[j]
                trocas += 1

    return vetor, comparacoes, trocas


dados = [5, 3, 8, 1, 2]

ordenado, comparacoes, trocas = bubble_sort(dados.copy())

print("Original:", dados)
print("Ordenado:", ordenado)
print("Comparações:", comparacoes)
print("Trocas:", trocas)


```
